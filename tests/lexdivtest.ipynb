```python
import nltk
# calcMetrique fonction qui prend un texte et retourne un dictionnaire avec les six métriques demandées
def calcMetrique(Text):
    tokens = nltk.word_tokenize(Text)
    clean_tokens = [t.lower() for t in tokens if t.isalpha()]
    types = set(clean_tokens)
    
    hapax = [t for t in types if clean_tokens.count(t) == 1]
    tok = len(clean_tokens)
    typ = len(types)
    hap = len(hapax)
    ttr = typ / tok if tok > 0 else 0
    mwl = sum(len(t) for t in clean_tokens) / tok if tok > 0 else 0
    mwf = tok / typ if typ > 0 else 0
    return {
      
        'tok': tok,
        'typ': typ,
        'hap': hap,
        'ttr': ttr,
        'mwl': mwl,
        'mwf': mwf,
        'clean_tokens': clean_tokens
    }
```

```python

text = "The cat sat on the mat. The cat is orange. The mat is brown. The cat and the mat are in the room."
print(calcMetrique(text))
```
