```python
import sys
import time
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

from nlp_pipeline.nlp import summarize
from nlp_pipeline.retrieval import get_metadata

BOOK = 11
```

```python
def resum_meta(book_id: int, *, use_cache: bool = True) -> str:
    """recup le resume gutendex depuis metadata"""
    meta = get_metadata(book_id, use_cache=use_cache)
    parts = meta.get("summaries") or []

    cleaned_parts = []

    for part in parts:
        if part:
            stripped_part = part.strip()

            if stripped_part:
                cleaned_parts.append(stripped_part)

    return " ".join(cleaned_parts)
```

```python
t0 = time.time()
res_gen = summarize.compute(BOOK, use_cache=False)
t_gen = time.time() - t0
res_ref = resum_meta(BOOK, use_cache=False)

print(f"livre {BOOK}")
print(f"temps {t_gen:.1f}s")
print("\n summary : ", res_gen)
print("\n ref : ", res_ref)
```
