```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import matplotlib.pyplot as plt
from nlp_pipeline.nlp.topic import topic_tfidf, topic_bow

chapitres_mots = {
    1: ['alice', 'rabbit', 'hole', 'fall', 'key', 'door', 'garden', 'bottle', 'drink', 'cake'],
    2: ['tears', 'pool', 'feet', 'fan', 'gloves', 'mabel', 'mouse', 'cat', 'swim', 'animals'],
    3: ['dry', 'dodo', 'caucus-race', 'prizes', 'comfits', 'thimble', 'tale', 'tail', 'fury', 'dinah'],
    4: ['house', 'mary ann', 'grow', 'arm', 'window', 'chimney', 'bill', 'pebbles', 'puppy', 'mushroom'],
    5: ['caterpillar', 'hookah', 'size', 'father william', 'mushroom', 'neck', 'serpent', 'pigeon', 'eggs', 'height'],
    6: ['footman', 'duchess', 'pepper', 'cook', 'baby', 'pig', 'cheshire cat', 'grin', 'mad', 'march hare'],
    7: ['tea-party', 'hatter', 'march hare', 'dormouse', 'room', 'riddle', 'watch', 'time', 'treacle', 'teapot'],
    8: ['roses', 'gardeners', 'queen', 'cards', 'beheaded', 'croquet', 'flamingo', 'hedgehog', 'king', 'executioner'],
    9: ['moral', 'mock turtle', 'gryphon', 'soup', 'school', 'sea', 'tortoise', 'lessons', 'reeling', 'uglification'],
    10: ['lobster', 'quadrille', 'dance', 'whiting', 'snail', 'song', 'sea-shore', 'mouth', 'shoes', 'france'],
    11: ['court', 'king', 'queen', 'knave', 'tarts', 'jury', 'witness', 'hatter', 'slate', 'evidence']
}
```

```python
res_tf = topic_tfidf.compute(11)
res_bw = topic_bow.compute(11)
```

```python
def norm_mots(mots):
    """normalise les mots"""
    tokens = []
    for m in mots:
        for part in m.lower().replace('-', ' ').split():
            tokens.append(part)
    return tokens

def overlap(a, b):
    sa, sb = set(norm_mots(a)), set(norm_mots(b))
    return len(sa & sb) / len(sa) if sa else 0

chs = sorted(chapitres_mots)
ovs_tf_opt, ovs_bw_opt, ovs = [], [], []

for c in chs:
    optimal = chapitres_mots[c]
    mots_tfidf = res_tf.get(c, [])
    mots_bow = res_bw.get(c, [])
    ovs_tf_opt.append(overlap(optimal, mots_tfidf))
    ovs_bw_opt.append(overlap(optimal, mots_bow))
    ovs.append(overlap(mots_tfidf, mots_bow))

print("proximite moyenne opti tfidf ", round(sum(ovs_tf_opt) / len(ovs_tf_opt), 2))
print("proximite moyenne opti bow ", round(sum(ovs_bw_opt) / len(ovs_bw_opt), 2))
print("ressemblance moyenne entre tfidf et bow ", round(sum(ovs) / len(ovs), 2))
```

```python
import numpy as np

positions_axes_x = np.arange(len(chs))
largeur_barre = 0.35

fig, ax = plt.subplots()
ax.bar(positions_axes_x - largeur_barre / 2, ovs_tf_opt, largeur_barre, label='tfidf')
ax.bar(positions_axes_x + largeur_barre / 2, ovs_bw_opt, largeur_barre, label='bow')
ax.set_xticks(positions_axes_x)
ax.set_xticklabels(chs)
ax.set_xlabel('document')
ax.set_ylabel('score à la réponse optimal')
ax.set_ylim(0, 1)
ax.legend()
plt.show()
```
