```python
import sys
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
sys.path.insert(0, str(Path.cwd().parent))

from nlp_pipeline.nlp.entities import spacy, spacy_md, nltk, personalise, eval

BOOK = 11
METHODES = {
    'spacy sm': spacy.compute,
    'spacy md': spacy_md.compute,
    'nltk': nltk.compute,
    'personalise': personalise.compute,
}
resultats = {nom: fn(BOOK, use_cache=False) for nom, fn in METHODES.items()}
scores = {nom: eval.compare(res) for nom, res in resultats.items()}
```

```python
noms = list(scores.keys())
print(f"{'methode':<14} {'pers':>8} {'lieu':>8}")
print('-' * 34)
for nom in sorted(noms, key=lambda n: -(scores[n]['characters']['score_equilibre'] + scores[n]['locations']['score_equilibre'])):
    s = scores[nom]
    print(f"{nom:<14} {s['characters']['score_equilibre']:>8.3f} {s['locations']['score_equilibre']:>8.3f}")
```

```python
x = np.arange(len(noms))
w = 0.35
fig, ax = plt.subplots()
ax.bar(x - w / 2, [scores[n]['characters']['score_equilibre'] for n in noms], w, label='pers')
ax.bar(x + w / 2, [scores[n]['locations']['score_equilibre'] for n in noms], w, label='lieu')
ax.set_xticks(x)
ax.set_xticklabels(noms, rotation=15)
ax.set_ylabel('score equilibre vs reference')
ax.set_ylim(0, 1)
ax.legend()
plt.tight_layout()
plt.show()
```
